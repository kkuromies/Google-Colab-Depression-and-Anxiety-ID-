BACA CSV

```python
import pandas as pd

# Membaca ketiga file CSV
df_rand = pd.read_csv('/content/datd_rand.csv')
df_test = pd.read_csv('/content/datd_test.csv')
df_train = pd.read_csv('/content/datd_train.csv')

# Menampilkan informasi singkat dari masing-masing DataFrame
print("Data Rand Shape:", df_rand.shape)
print("Data Test Shape:", df_test.shape)
print("Data Train Shape:", df_train.shape)
```

HITUNG JUMLAH DATA PADA CSV BERIKUT DENGAN PROPORSI LABEL 1

```python
import pandas as pd

# List dataframes and their names
dataframes = {
    'datd_rand.csv': df_rand,
    'datd_test.csv': df_test,
    'datd_train.csv': df_train
}

summary_data = []

for name, df in dataframes.items():
    # Menghitung jumlah label 0 dan label 1
    count_0 = (df['label'] == 0).sum()
    count_1 = (df['label'] == 1).sum()
    total = len(df)

    # Menghitung persentase label 1
    pct_1 = (count_1 / total) * 100 if total > 0 else 0

    summary_data.append({
        'Nama CSV': name,
        'Jumlah Label 0': count_0,
        'Jumlah Label 1': count_1,
        'Jumlah Seluruh Data': total,
        'Persentase Label 1 (%)': round(pct_1, 2)
    })

# Membuat DataFrame ringkasan
df_summary = pd.DataFrame(summary_data)
df_summary
```

AUDIT DUPLIKAT BARIS,LABEL KONTRADIKTIF DAN KEBOCORAN DATA

```python
import pandas as pd

# 1. Pemeriksaan Duplikasi Baris
print("=== PEMERIKSAAN DUPLIKASI BARIS ===")
for name, df in dataframes.items():
    duplicates = df.duplicated().sum()
    print(f"{name}: ditemukan {duplicates} baris duplikat secara keseluruhan.")

# 2. Pemeriksaan Label Kontradiktif (Teks sama tetapi label berbeda)
print("\n=== PEMERIKSAAN LABEL KONTRADIKTIF ===")
for name, df in dataframes.items():
    # Kelompokkan berdasarkan teks dan cari teks yang memiliki lebih dari satu label unik
    contradictory = df.groupby('text')['label'].nunique()
    contradictory_texts = contradictory[contradictory > 1]
    print(f"{name}: ditemukan {len(contradictory_texts)} teks dengan label kontradiktif.")
    if len(contradictory_texts) > 0:
        print("Contoh teks kontradiktif:")
        print(df[df['text'].isin(contradictory_texts.index)].sort_values(by='text').head(4))

# 3. Pemeriksaan Kebocoran Data (Data uji bocor ke data latih)
print("\n=== PEMERIKSAAN KEBOCORAN DATA (TEST -> TRAIN) ===")
# Irisan teks antara data latih dan data uji
train_texts = set(df_train['text'])
test_texts = set(df_test['text'])
leakage_texts = train_texts.intersection(test_texts)

print(f"Jumlah teks dari data uji yang bocor ke data latih: {len(leakage_texts)} teks.")
if len(leakage_texts) > 0:
    print("Contoh data uji yang bocor ke data latih:")
    print(df_test[df_test['text'].isin(list(leakage_texts)[:3])])
```
